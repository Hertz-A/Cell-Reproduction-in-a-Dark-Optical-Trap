```python
import numpy as np
import matplotlib.pyplot as plt

FPS    = 30
PX2UM  = 0.053
CONF   = 0.7

# ---------------------------------------------------------------- filtragem
raw = {1: df_1, 2: df_2, 4: df_4, 5: df_5, 6: df_6,
       7: df_7, 8: df_8, 9: df_9, 11: df_11, 12: df_12}   # df_3 e df_10 descartados

roi = {1: (800, 1100, 500),   2: (800, 1100, 500),
       4: (900, 1050, 550),   5: (800, 1150, 600),
       6: (800, 1150, 600),   7: (800, 1150, 600),
       8: (800, 1150, 600),   9: (800, 1100, 500),
      11: (800, 1100, 500),  12: (800, 1100, 500)}

free_point = {1: 1832, 2: 1813, 4: 1825, 5: 1807,  6: 1790,
              7: 1806, 8: 1810, 9: 1800, 11: 1800, 12: 1800}

dfs = {}
for k, df in raw.items():
    xmin, xmax, ymax = roi[k]
    mask = ((df["Centro_X"] >= xmin) & (df["Centro_X"] <= xmax)
            & (df["Centro_Y"] <= ymax) & (df["Confiança"] > CONF))
    dfs[k] = df.loc[mask]

print({k: len(v) for k, v in dfs.items()})

# ---------------------------------------------------------------- figura
fontSize = 15
fig, axs = plt.subplots(2, figsize=(7, 5))

for k, df in dfs.items():
    frames  = df.index.values            # frame original, preservado pelo .loc
    t       = frames / FPS
    trapped = frames < free_point[k]     # máscara pelo frame, não pela posição

    xs = df["Centro_X"].values
    ys = df["Centro_Y"].values
    x  = (xs - xs[trapped].mean()) * PX2UM
    y  = (ys - ys[trapped].mean()) * PX2UM

    axs[0].plot(t, x, linewidth=0.4, label=f"Cell {k}")
    axs[1].plot(t, y, linewidth=0.4, label=f"Cell {k}")

t_free = np.mean(list(free_point.values())) / FPS
t_end  = max(df.index.max() for df in dfs.values()) / FPS

for ax, lab in zip(axs, [r"$x$ [$\mu$m]", r"$y$ [$\mu$m]"]):
    ax.set_xlabel(r"Time [$s$]", fontsize=fontSize)
    ax.set_ylabel(lab, fontsize=fontSize)
    ax.set_ylim(-50 * PX2UM, 50 * PX2UM)
    ax.axvspan(t_free, t_end, color="lightblue", alpha=0.5)
    ax.tick_params(axis="both", labelsize=14)
    ax.grid()

plt.tight_layout()
# plt.savefig("plots_hold2free.pdf", format="pdf", dpi=2000, bbox_inches="tight")
plt.show()
```

```python
from scipy.optimize import curve_fit

fontSize = 20

dfs = [df_1, df_2, df_4, df_5, df_6, df_7, df_8, df_9, df_11, df_12]
indices = [1, 2, 4, 5, 6, 7, 8, 9, 11, 12]

fps = 30
frame_split = 60 * fps  # segundo 60
n_bins = 60

all_x_before, all_y_before = [], []
all_x_after, all_y_after = [], []

for i, df in zip(indices, dfs):
    mean_x = np.mean(df['Centro_X'][:frame_split])
    mean_y = np.mean(df['Centro_Y'][:frame_split])

    x_centered = (df['Centro_X'].values - mean_x) * 0.053
    y_centered = (df['Centro_Y'].values - mean_y) * 0.053

    all_x_before.append(x_centered[:frame_split])
    all_y_before.append(y_centered[:frame_split])
    all_x_after.append(x_centered[frame_split:])
    all_y_after.append(y_centered[frame_split:])

all_x_before = np.concatenate(all_x_before)
all_y_before = np.concatenate(all_y_before)
all_x_after = np.concatenate(all_x_after)
all_y_after = np.concatenate(all_y_after)

bins_x = np.linspace(min(all_x_before.min(), all_x_after.min()),
                      max(all_x_before.max(), all_x_after.max()), n_bins)
bins_y = np.linspace(min(all_y_before.min(), all_y_after.min()),
                      max(all_y_before.max(), all_y_after.max()), n_bins)


def gaussian(x, amplitude, mu, sigma):
    return amplitude * np.exp(-(x - mu) ** 2 / (2 * sigma ** 2))


def plot_hist_with_gaussian(ax, data, bins, color, label):
    counts, edges = np.histogram(data, bins=bins, density=True)
    centers = (edges[:-1] + edges[1:]) / 2

    ax.hist(data, bins=bins, density=True, color=color, edgecolor='black', alpha=0.5, label=label)

    p0 = [counts.max(), centers[np.argmax(counts)], np.std(data) / 2]
    popt, _ = curve_fit(gaussian, centers, counts, p0=p0, maxfev=5000)

    x_fit = np.linspace(-3, 3, 300)
    ax.plot(x_fit, gaussian(x_fit, *popt), color=color, linewidth=2)


fig, axs = plt.subplots(1, 2, figsize=(9.5, 4.5),sharey=True)

plot_hist_with_gaussian(axs[0], all_x_before, bins_x, 'tab:blue', 'Trapped')
plot_hist_with_gaussian(axs[0], all_x_after, bins_x, 'tab:orange', 'Free')
axs[0].set_xlabel(r'$x$ [$\mu$m]', fontsize=fontSize)
axs[0].set_ylabel('Probability density', fontsize=fontSize)
axs[0].tick_params(axis='both', labelsize=18)
axs[0].grid(alpha=0.3)
axs[0].legend(fontsize=16)
axs[0].set_ylim(0,5.3)
axs[0].set_xlim(-2.65,2.65)

plot_hist_with_gaussian(axs[1], all_y_before, bins_y, 'tab:blue', 'Trapped')
plot_hist_with_gaussian(axs[1], all_y_after, bins_y, 'tab:orange', 'Free')
axs[1].set_xlabel(r'$y$ [$\mu$m]', fontsize=fontSize)
axs[1].tick_params(axis='both', labelsize=18)
axs[1].grid(alpha=0.3)
axs[1].legend(fontsize=16)
axs[1].set_ylim(0,5.3)
axs[1].set_xlim(-2.65,2.65)

plt.tight_layout()
plt.savefig("position_distribution_before_after.pdf", format="pdf", bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

FPS, PX2UM, CONF, NBINS = 30, 0.053, 0.7, 60
YLIM = 50 * PX2UM          # 2.65 um

# ---------------------------------------------------------------- filtragem
raw = {1: df_1, 2: df_2, 4: df_4, 5: df_5, 6: df_6,
       7: df_7, 8: df_8, 9: df_9, 11: df_11, 12: df_12}

roi = {1: (800, 1100, 500),   2: (800, 1100, 500),
       4: (900, 1050, 550),   5: (800, 1150, 600),
       6: (800, 1150, 600),   7: (800, 1150, 600),
       8: (800, 1150, 600),   9: (800, 1100, 500),
      11: (800, 1100, 500),  12: (800, 1100, 500)}

free_point = {1: 1832, 2: 1813, 4: 1825, 5: 1807,  6: 1790,
              7: 1806, 8: 1810, 9: 1800, 11: 1800, 12: 1800}

dfs = {}
for k, df in raw.items():
    xmin, xmax, ymax = roi[k]
    dfs[k] = df.loc[(df["Centro_X"] >= xmin) & (df["Centro_X"] <= xmax)
                    & (df["Centro_Y"] <= ymax) & (df["Confiança"] > CONF)]

# ---------------------------------------------------------------- séries
traces = {}
for k, df in dfs.items():
    frames  = df.index.values
    trapped = frames < free_point[k]
    xs, ys  = df["Centro_X"].values, df["Centro_Y"].values
    traces[k] = dict(t=frames / FPS,
                     x=(xs - xs[trapped].mean()) * PX2UM,
                     y=(ys - ys[trapped].mean()) * PX2UM,
                     trapped=trapped)

pool = lambda c, m: np.concatenate([tr[c][tr["trapped"] == m] for tr in traces.values()])
x_trap, x_free = pool("x", True), pool("x", False)
y_trap, y_free = pool("y", True), pool("y", False)

bins = np.linspace(-YLIM, YLIM, NBINS)

# ---------------------------------------------------------------- ajuste
def gaussian(v, a, mu, sigma):
    return a * np.exp(-(v - mu) ** 2 / (2 * sigma ** 2))

def hist_h(ax, data, color, label):
    counts, edges = np.histogram(data, bins=bins, density=True)
    centers = (edges[:-1] + edges[1:]) / 2
    ax.hist(data, bins=bins, density=True, orientation="horizontal",
            color=color, edgecolor="black", alpha=0.5, label=label)
    popt, _ = curve_fit(gaussian, centers, counts, maxfev=5000,
                        p0=[counts.max(), centers[np.argmax(counts)], np.std(data) / 2])
    v = np.linspace(bins[0], bins[-1], 300)
    ax.plot(gaussian(v, *popt), v, color=color, linewidth=2)
    return popt

# ---------------------------------------------------------------- figura
fs = 15
fig, axs = plt.subplots(2, 2, figsize=(11, 6), sharex="col", sharey="row",
                        gridspec_kw={"width_ratios": [3.2, 1]})

for k, tr in traces.items():
    axs[0, 0].plot(tr["t"], tr["x"], linewidth=0.4)
    axs[1, 0].plot(tr["t"], tr["y"], linewidth=0.4)

sig = {}
sig["x_trap"] = hist_h(axs[0, 1], x_trap, "tab:blue",   "Trapped")
sig["x_free"] = hist_h(axs[0, 1], x_free, "tab:orange", "Free")
sig["y_trap"] = hist_h(axs[1, 1], y_trap, "tab:blue",   "Trapped")
sig["y_free"] = hist_h(axs[1, 1], y_free, "tab:orange", "Free")

t_free = np.mean(list(free_point.values())) / FPS
t_end  = max(df.index.max() for df in dfs.values()) / FPS

for row, lab in zip(range(2), [r"$x$ [$\mu$m]", r"$y$ [$\mu$m]"]):
    axs[row, 0].set_ylabel(lab, fontsize=fs)
    axs[row, 0].set_ylim(-YLIM, YLIM)
    axs[row, 0].axvspan(t_free, t_end, color="lightblue", alpha=0.5)
    axs[row, 0].grid(alpha=0.3)
    axs[row, 1].grid(alpha=0.3)
    axs[row, 1].set_xlim(0, 5.3)
    for col in range(2):
        axs[row, col].tick_params(axis="both", labelsize=13)

axs[1, 0].set_xlabel(r"Time [$s$]", fontsize=fs)
axs[1, 1].set_xlabel("Probability density", fontsize=fs)
axs[0, 1].legend(fontsize=12, loc="upper right")

#for ax, tag in zip(axs.ravel(), ["(a)", "(b)", "(c)", "(d)"]):
#    ax.text(0.02, 0.94, tag, transform=ax.transAxes,
#            fontsize=fs, fontweight="bold", va="top")

plt.tight_layout()
plt.savefig("confinement.pdf", format="pdf", bbox_inches="tight")
plt.show()

for key, p in sig.items():
    print(f"{key}: sigma = {abs(p[2]):.3f} um")
```

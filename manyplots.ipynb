```python
eta1 = 11.1 #mPa
eta2 = 0.69 #mPa

#def viscosity(w): 
#    return (((w)/eta1)+((1-w)/eta2))**(-1)

def viscosity(w): # definição de Arrhenius para viscosidade
    return np.exp(w*np.log(eta1) + (1-w)*np.log(eta2))
```

```python
excluir = {
    "dfs_sol1": [2,5],
    "dfs_sol2": [2],
    "dfs_sol3": [3,7],
    "dfs_sol4": [3,4],
    "dfs_sol5": [0],
    "dfs_sol6": [1, 4, 6, 8],
    "dfs_sol7": [2,9]
}

concentrations = [(1*0.6)/(1.03), (0.95*0.6)/(1.03), (0.9*0.6)/(1.03), (0.85*0.6)/(1.03), (0.8*0.6)/(1.03), (0.75*0.6)/(1.03), (0.7*0.6)/(1.03)]

viscosities = np.array([viscosity(w) for w in concentrations])

mean_variances_x = []
mean_variances_y = []
mean_stds_x = []
mean_stds_y = []

dfs_super = [dfs_sol1, dfs_sol2, dfs_sol3, dfs_sol4, dfs_sol5, dfs_sol6, dfs_sol7]

for i, dfs in enumerate(dfs_super):
    group_name = f"dfs_sol{i+1}"
    
    variances_x = []
    variances_y = []
    stds_x = []
    stds_y = []
    
    for j, df in enumerate(dfs):
        if j not in excluir.get(group_name, []):
            diameter_mean = df["Diâmetro"].mean()
            var_x = df["Centro_X"].var() / (diameter_mean * viscosities[i])
            var_y = df["Centro_Y"].var() / (diameter_mean * viscosities[i])
            std_x = df["Centro_X"].std() / (diameter_mean * viscosities[i])
            std_y = df["Centro_Y"].std() / (diameter_mean * viscosities[i])

            variances_x.append(var_x)
            variances_y.append(var_y)
            stds_x.append(std_x)
            stds_y.append(std_y)
    
    mean_variances_x.append(np.mean(variances_x))
    mean_variances_y.append(np.mean(variances_y))
    
    mean_stds_x.append(np.mean(stds_x))
    mean_stds_y.append(np.mean(stds_y))

mean_variances_x = np.array(mean_variances_x)
mean_variances_y = np.array(mean_variances_y)

mean_stds_x = np.array(mean_stds_x)
mean_stds_y = np.array(mean_stds_y)

concentrations = np.array(concentrations)
```

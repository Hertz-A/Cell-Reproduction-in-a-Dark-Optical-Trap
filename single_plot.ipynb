```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.cm as cm
```

```python
csv_path = r"D:\ClawLike Tweezer to Biology\variance measurement\solution 5\data in csv - solution 5\sol5 - v6 - yolo.csv"
```

```python
df = pd.read_csv(csv_path, encoding='cp1252')

#df = df.loc[df["Centro_X"] >= 800]
#df = df.loc[df["Centro_X"] <= 500]
#df = df.loc[df["Centro_Y"] <= 500]
df = df.loc[df["Confiança"] > 0.7]

len(df)
```

```python
fontSize = 15
fig, axs = plt.subplots(2)
fig.set_figwidth(8)
fig.set_figheight(6)
fig.suptitle(r'Coordinates through time', fontsize=18)
plt.subplots_adjust(left=0.1,
                    bottom=0.1,
                    right=0.9,
                    top=0.9,
                    wspace=0.4,
                    hspace=0.4)

frame2seconds = np.arange(len(df)) / 30
mean_x = np.mean(df['Centro_X'])
mean_y = np.mean(df['Centro_Y'])
std_x = np.std(df['Centro_X'])
std_y = np.std(df['Centro_Y'])
axs[0].plot(frame2seconds, (df['Centro_X'].values - mean_x)*0.053, linewidth=0.4, label=f'Curve')
axs[1].plot(frame2seconds, (df['Centro_Y'].values - mean_y)*0.053, linewidth=0.4, label=f'Curve')

axs[0].set_xlabel(r'Time [$s$]', fontsize=fontSize)
axs[0].set_ylabel(r'$x$ [$\mu$m]', fontsize=fontSize)
axs[0].set_ylim(-50*0.053, 50*0.053)
axs[0].grid()

axs[1].set_xlabel(r'Time [$s$]', fontsize=fontSize)
axs[1].set_ylabel(r'$y$ [$\mu$m]', fontsize=fontSize)
axs[1].set_ylim(-50*0.053, 50*0.053)
axs[1].grid()

axs[0].axvspan(1800/30, 3640/30, color='lightblue', alpha=0.5)
axs[1].axvspan(1800/30, 3640/30, color='lightblue', alpha=0.5)

axs[0].tick_params(axis='both', labelsize=14)
axs[1].tick_params(axis='both', labelsize=14)

plt.tight_layout()
plt.show()
```

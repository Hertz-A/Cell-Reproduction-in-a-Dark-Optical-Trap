```python
from scipy.signal import savgol_filter

def smooth(data, window_frac=0.055, polyorder=3):
    window = int(len(data) * window_frac)
    if window % 2 == 0:
        window += 1
    window = max(window, polyorder + 2)
    return savgol_filter(data, window_length=window, polyorder=polyorder)

plt.figure(figsize=(7.5, 5))
fontSize = 20

t4h = np.arange(len(df_mean4h)) * 0.0344 * 0.25
t1h = np.arange(len(df_mean1h)) * 0.0344 * 0.25
t10min = np.arange(len(df_mean10min)) * 0.0344 * 0.25
tg10min = np.arange(len(df_mean_gauss)) * 0.03521 * 0.25

plt.plot(tg10min, smooth(df_mean_gauss), label="gaussian, 10min")
plt.plot(t4h, smooth(df_mean4h), label="4h trapped")
plt.plot(t1h, smooth(df_mean1h), label="1h trapped")
plt.plot(t10min, smooth(df_mean10min), label="10 min trapped")

plt.axhline(y=1, color='black', linestyle='--', linewidth=1, alpha=1)
plt.axhline(y=2, color='black', linestyle='--', linewidth=1, alpha=1)

plt.ylim(-0.1, 2.1)
plt.xlim(-0.1, 4.1)

plt.xlabel(r'Time [$h$]', fontsize=fontSize)
plt.ylabel(r'Total area', fontsize=fontSize)
plt.tick_params(axis='both', labelsize=18)

plt.grid(True)
plt.legend(fontsize=16, loc='lower right')
plt.tight_layout()
#plt.savefig("growing_analysis.pdf", format="pdf", dpi=2000, bbox_inches="tight")
plt.show()
```

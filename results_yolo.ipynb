```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('results.csv')
```

```python
import matplotlib.pyplot as plt

title_size = 16
label_size = 20
tick_size = 20

#plt.rcParams['font.family'] = 'serif'

colunas = ['metrics/mAP50(B)', 'metrics/mAP50-95(B)']
nomes_legais = {
    'metrics/mAP50(B)': 'mAP@50',
    'metrics/mAP50-95(B)': 'mAP@50-95',
}

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(11, 4))

for ax, coluna in zip(axes, colunas):
    ax.plot(df['epoch'], df[coluna], color='tab:blue', marker='o')#, linewidth=1.8)
    #ax.set_title(nomes_legais[coluna], fontsize=title_size)
    ax.set_xlabel('Epoch', fontsize=label_size)
    ax.set_ylabel(nomes_legais[coluna], fontsize=label_size)
    ax.tick_params(axis='both', labelsize=tick_size)
    ax.grid(True, alpha=0.3)
    ax.set_ylim(-0.05, 1.05)

plt.tight_layout()
plt.savefig("yolo_metrics.pdf", format='pdf', bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt

label_size = 16
tick_size = 16
legend_size = 16

#plt.rcParams['font.family'] = 'serif'

colunas = ['metrics/mAP50(B)', 'metrics/mAP50-95(B)']
nomes_legais = {
    'metrics/mAP50(B)': 'mAP@50',
    'metrics/mAP50-95(B)': 'mAP@50-95',
}
cores = {
    'metrics/mAP50(B)': 'tab:blue',
    'metrics/mAP50-95(B)': 'tab:orange',
}

fig, ax = plt.subplots(figsize=(7, 5))

for coluna in colunas:
    ax.plot(df['epoch'], df[coluna], color=cores[coluna], marker='o', label=nomes_legais[coluna])

ax.set_xlabel('Epoch', fontsize=label_size)
ax.set_ylabel('Value', fontsize=label_size)
ax.tick_params(axis='both', labelsize=tick_size)
ax.grid(True, alpha=0.3)
ax.set_ylim(-0.05, 1.05)
ax.legend(fontsize=legend_size)

plt.tight_layout()
#plt.savefig("yolo_metrics.pdf", format='pdf', bbox_inches='tight')
plt.show()
```

```python


fontSize = 15
dfs_sol7 = [df_sol7_1, df_sol7_2, df_sol7_4, df_sol7_5, df_sol7_6, df_sol7_7, df_sol7_8]

# Definição das soluções e seus rótulos
dfs_super = [dfs_sol7]
sol_labels = ["solution 7"]

# Determinar o tempo máximo real baseado nos dados
max_time = max(len(df) for dfs in dfs_super for df in dfs) / 30
free_point = {1: 1832, 2: 1813, 3: 1838, 4: 1825, 5: 1807, 6: 1790, 7: 1806, 8: 1810}

for sol_label, dfs in zip(sol_labels, dfs_super):
    for i, df in enumerate(dfs, 1):
        fig, axs = plt.subplots(2, figsize=(8, 6), sharex=True)
        fig.suptitle(f'Coordinates through time - {sol_label}, video {i}', fontsize=18)

        frame2seconds = np.arange(len(df)) / 30
        n = free_point[i]

        mean_x = np.mean(df['Centro_X'][:n])
        mean_y = np.mean(df['Centro_Y'][:n])
        std_x = np.std(df['Centro_X'][:n])
        std_y = np.std(df['Centro_Y'][:n])

        line_x, = axs[0].plot([], [], linewidth=0.4, label=f'Curve {i}')
        line_y, = axs[1].plot([], [], linewidth=0.4, label=f'Curve {i}')

        # Definir limites do eixo x e y automaticamente
        axs[0].set_xlim(0, max_time)
        axs[1].set_xlim(0, max_time)
        axs[0].set_ylim(-20 * 0.053, 20 * 0.053)
        axs[1].set_ylim(-20 * 0.053, 20 * 0.053)

        axs[0].set_xlabel(r'Time [$s$]', fontsize=fontSize)
        axs[0].set_ylabel(r'$x$ [$μm$]', fontsize=fontSize)
        axs[0].grid()

        axs[1].set_xlabel(r'Time [$s$]', fontsize=fontSize)
        axs[1].set_ylabel(r'$y$ [$μm$]', fontsize=fontSize)
        axs[1].grid()

        axs[0].tick_params(axis='both', labelsize=14)
        axs[1].tick_params(axis='both', labelsize=14)

        def update(frame):
            line_x.set_data(frame2seconds[:frame], (df['Centro_X'].values[:frame] - mean_x) * 0.053)
            line_y.set_data(frame2seconds[:frame], (df['Centro_Y'].values[:frame] - mean_y) * 0.053)
            return line_x, line_y

        ani = animation.FuncAnimation(fig, update, frames=len(df), interval=1000/30, blit=False)

        filename = f"{sol_label} - video {i} - yolo.gif"
        #ani.save(filename, writer="pillow", fps=30)

        plt.close(fig)
```
